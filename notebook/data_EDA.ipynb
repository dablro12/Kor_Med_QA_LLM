## aihub_전문_의학지식_데이터

```python
train_root_dir = "/mnt/sdd/kor_med_opendataset/aihub_전문_의학지식_데이터/Training/02.라벨링데이터"
valid_root_dir = "/mnt/sdd/kor_med_opendataset/aihub_전문_의학지식_데이터/Validation/02.라벨링데이터"
```

```python
from glob import glob 
import pandas as pd
from pathlib import Path
from collections import defaultdict
import json

train_files = glob(train_root_dir + "/**/*.json")
valid_files = glob(valid_root_dir + "/**/*.json")

# 과별 파일 개수 집계
train_dept_counts = defaultdict(int)
valid_dept_counts = defaultdict(int)

for file_path in train_files:
    # 파일 경로에서 과 이름 추출 (TL_제거)
    dept_name = Path(file_path).parent.name.replace("TL_", "")
    train_dept_counts[dept_name] += 1

for file_path in valid_files:
    # 파일 경로에서 과 이름 추출 (VL_제거)
    dept_name = Path(file_path).parent.name.replace("VL_", "")
    valid_dept_counts[dept_name] += 1

# 모든 과 이름 수집
all_depts = sorted(set(list(train_dept_counts.keys()) + list(valid_dept_counts.keys())))

# DataFrame 생성
df = pd.DataFrame({
    '과': all_depts,
    'Train 파일 수': [train_dept_counts[dept] for dept in all_depts],
    'Valid 파일 수': [valid_dept_counts[dept] for dept in all_depts]
})

# 총계 행 추가
df.loc[len(df)] = ['총계', df['Train 파일 수'].sum(), df['Valid 파일 수'].sum()]

print("#"* 15 + "aihub_전문_의학지식_데이터" + "#"* 15)
print(f"train_files: {len(train_files)}개")
print(f"valid_files: {len(valid_files)}개")
print("\n과별 파일 개수:")
print(df)

# 파일 하나 열어서 구조 확인
if train_files:
    sample_file = train_files[0]
    print(f"\n샘플 파일: {sample_file}")
    with open(sample_file, 'r', encoding='utf-8-sig') as f:
        sample_data = json.load(f)
    print("\n파일 구조:")
    print(json.dumps(sample_data, ensure_ascii=False, indent=2))
```

## sean0042_KorMedMCQA

```python
import pandas as pd
from pathlib import Path

root_dir = "/mnt/sdd/kor_med_opendataset/sean0042_KorMedMCQA"
categories = ['dentist', 'doctor', 'nurse', 'pharm']
file_types = ['train', 'dev', 'test', 'fewshot']

# 데이터 개수 집계
data_counts = []

for category in categories:
    for file_type in file_types:
        file_path = Path(root_dir) / category / f"{category}_{file_type}.csv"
        if file_path.exists():
            df_temp = pd.read_csv(file_path)
            data_counts.append({
                '카테고리': category,
                '파일 타입': file_type,
                '데이터 개수': len(df_temp)
            })
        else:
            data_counts.append({
                '카테고리': category,
                '파일 타입': file_type,
                '데이터 개수': 0
            })

# DataFrame 생성
df = pd.DataFrame(data_counts)

# 피벗 테이블로 변환 (카테고리를 행으로, 파일 타입을 열로)
df_pivot = df.pivot(index='카테고리', columns='파일 타입', values='데이터 개수')
df_pivot = df_pivot.fillna(0).astype(int)

# 총계 행 추가
df_pivot.loc['총계'] = df_pivot.sum()

print("#"* 15 + "sean0042_KorMedMCQA" + "#"* 15)
print("\n카테고리별 데이터 개수:")
display(df_pivot)

# 샘플 파일 하나 열어서 구조 확인
sample_file = None
for category in categories:
    for file_type in file_types:
        file_path = Path(root_dir) / category / f"{category}_{file_type}.csv"
        if file_path.exists():
            sample_file = file_path
            break
    if sample_file:
        break

if sample_file:
    print(f"\n샘플 파일: {sample_file}")
    df_sample = pd.read_csv(sample_file)
    print(f"\n컬럼: {', '.join(df_sample.columns.tolist())}")
    print(f"\n샘플 데이터 (첫 번째 행):")
    display(df_sample.head(1))
```

## snuh_ClinicalQA

```python
import pandas as pd

file_path = "/mnt/sdd/kor_med_opendataset/snuh_ClinicalQA/train.csv"
df = pd.read_csv(file_path)

# 기본 정보
print("#"* 15 + "snuh_ClinicalQA" + "#"* 15)
print(f"전체 데이터 개수: {len(df)}개")
print(f"\n컬럼: {', '.join(df.columns.tolist())}")

# 카테고리별 데이터 개수 (category 컬럼이 있는 경우)
if 'category' in df.columns:
    category_counts = df['category'].value_counts().reset_index()
    category_counts.columns = ['카테고리', '데이터 개수']
    category_counts = category_counts.sort_values('카테고리')
    
    # 총계 행 추가
    category_counts.loc[len(category_counts)] = ['총계', category_counts['데이터 개수'].sum()]
    
    print("\n카테고리별 데이터 개수:")
    print(category_counts)

# 데이터 미리보기
print("\n데이터 미리보기 (상위 3개):")
display(df.head(3))
```

## snuh_KorMedLawQA


```python
import pandas as pd
import json

file_path = "/mnt/sdd/kor_med_opendataset/snuh_KorMedLawQA/medical_law_qa_dataset.jsonl"

# JSONL 파일 읽기
data = []
with open(file_path, 'r', encoding='utf-8') as f:
    for line in f:
        if line.strip():
            data.append(json.loads(line))

df = pd.DataFrame(data)

# 기본 정보
print("#"* 15 + "snuh_KorMedLawQA" + "#"* 15)
print(f"전체 데이터 개수: {len(df)}개")
print(f"\n컬럼: {', '.join(df.columns.tolist())}")

# 법률별 데이터 개수 (law_title 컬럼이 있는 경우)
if 'law_title' in df.columns:
    law_counts = df['law_title'].value_counts().reset_index()
    law_counts.columns = ['법률', '데이터 개수']
    law_counts = law_counts.sort_values('법률')
    
    # 총계 행 추가
    law_counts.loc[len(law_counts)] = ['총계', law_counts['데이터 개수'].sum()]
    
    print("\n법률별 데이터 개수:")
    print(law_counts)

# 데이터 미리보기
print("\n데이터 미리보기 (상위 3개):")
display(df.head(3))
```
